```python
import math
```

Seguimos los pasos libro Burden de la pagina 161, algoritmo 4.1. donde aparece este pseudocódigo

```python
def romberbeg (f,a,b,n):
  if n<=0:
    ValueError("n debe ser mayor a 0")
  R1=[0.0] * (n+1)
  R2=[0.0] * (n+1)

  h=b-a
  R1[1]=(h/2.0) * (f(a)+f(b))

  print (f"R[1,1]={R1[1]:.88f}")
  for i in range (2,n+1):
    suma=0.0
    limite=2**(i-2)
    for k in range(1,limite+1):
      suma+= f(a+(k-0.5)*h)

    R2[1]= 0.5*(R1[1]+h*suma)

    for j in range(2,i+1):
      R2[j]=R2[j-1] + (R2[j-1]-R1[j-1])/(4**(j-1)-1)

    fila= [f"R[{i}, {j}] = {R2[j]:.8f}" for j in range(1, i + 1)]
    print(" | ".join(fila))

    h=h/2.0
    for j in range (1,i+1):
      R1[j] = R2[2]
  return R1[n]
```

Definimos la funcion del ejemplo 2 de la pagina 160 del Burden: $\int_0^\pi \sin x\,dx$, con a=0, b=$\pi$ y n=2


```python
def f(x):
  return math.sin(x)
```

```python
a=0
b=math.pi
n=2
```

Así se llega a las respuestas que aparecen en el Burden

```python
resultado= romberbeg (f,a,b,n)
print(resultado)
```
