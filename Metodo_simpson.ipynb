```python
import math
```

Seguimos el pseudocódigo de el libro de Burden, algoritmo 4.1, página 153

```python
def simpson (f, a, b, n):
  if n%2!=0 or n<= 0:
     raise ValueError("n debe ser entero positivo par")
  h= (b-a) / n
  XI0 = f(a) + f(b)
  XI1 = 0.0
  XI2 = 0.0

  for i in range(1, n):
    X = a + i*h
    if i%2==0:
      XI2+= f(X)
    else:
      XI1+= f(X)
  XI = h*(XI0+2 * XI2+4 * XI1)/3
  return XI
```

Definimos la funcion de la tarea 1: $\int_{0}^{2} e^{2x} \sin(3x) \, dx$, con a=0 y b=0 y n=12

```python
def f(x):
  return math.exp(2*x)*math.sin(3*x)
```

```python
a=0
b=2
n=12
```

De esta forma llamamos a la funcion creada en pasos anteriores con estos parámetros, dando como resultado -14.208518602994692

```python
resultado = simpson(f,a,b,n)
print (resultado )
```
