<h1 style="color: #2E86C1; text-align: center;">Cuadratura adaptable</h1>



```python
import math
```

Seguimos los pasos del pseudodedigo del libro de Burden (Algoritmo 4.3, página 166)

```python
def cuadratura (f,a,b,tol,n):
  tammax= 2*n+2
  a_arr = [0.0]*tammax
  h_arr = [0.0]*tammax
  FA_arr = [0.0]*tammax
  FC_arr = [0.0]*tammax
  FB_arr = [0.0]*tammax
  S_arr  = [0.0]*tammax
  tol_arr = [0.0]*tammax
  L_arr = [0]*tammax
  APP = 0.0
  i=1
  tol_arr[i] = 10.0*tol
  a_arr[i]   = a
  h_arr[i]   = (b-a)/2.0
  FA_arr[i]  = f(a)
  FC_arr[i]  = f(a+h_arr[i])
  FB_arr[i]  = f(b)
  S_arr[i]   = h_arr[i]*(FA_arr[i]+4.0*FC_arr[i]+FB_arr[i])/3.0
  L_arr[i]   = 1
  while i > 0:
    FD = f(a_arr[i] + h_arr[i] / 2.0)
    FE = f(a_arr[i] + 3.0 * h_arr[i] / 2.0)
    S1 = h_arr[i] * (FA_arr[i] + 4.0*FD + FC_arr[i])/6.0
    S2 = h_arr[i] * (FC_arr[i] + 4.0*FE + FB_arr[i])/6.0
    v_1=a_arr[i]
    v_2=FA_arr[i]
    v_3=FC_arr[i]
    v_4=FB_arr[i]
    v_5=h_arr[i]
    v_6=tol_arr[i]
    v_7=S_arr[i]
    v_8=L_arr[i]

    i=-1
    if abs(S1+S2-v_7)<=v_6:
      APP+=(S1+S2)
    else:
      if v_8>=n:
        return ("nivel excedido")
      else:
        i+=1
        a_arr[i]=v_1+v_5
        FA_arr[i]=v_3
        FC_arr[i]=FE
        FB_arr[i]=v_4
        h_arr[i]=v_5/2.0
        tol_arr[i]=v_6/2.0
        S_arr[i]=S2
        L_arr[i]=v_8+1

        i+=1
        a_arr[i]=v_1
        FA_arr[i]=v_2
        FC_arr[i]=FD
        FB_arr[i]=v_3
        h_arr[i]=h_arr[i-1]
        tol_arr[i]=tol_arr[i-1]
        S_arr[i]=S1
        L_arr[i]=L_arr[i-1]
  return APP

```

Definimos la funcion de la ilustración de la página 167 del Burden f(x) = $(\frac{100}{x^2}) \operatorname{sen}(\frac{10}{x})$ con:
a=1
b=3
una tolerancia de $10^{-4}$
n=220

```python
def f(x):
  return (100/x**2)*math.sin(10/x)
```

```python
a=1.0
b=3.0
tol=0.0001
n=20
```

Con esto nos da el resultado de: -1.241123522553495

```python
resultado=cuadratura(f,a,b,tol,n)
print(resultado)
```
